```python
import cv2 as cv
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm
import numpy as np
import glob
```

```python
#metodo para mostrar brillo y oscuro
def mostrar_dos_imagenes(brillo, oscuro, titulos=("Brillo", "Oscuro")):
    imagenes = [brillo, oscuro]
    fig, axes = plt.subplots(1,2, figsize=(5,6))
    for(ax, img, titulo) in zip(axes, imagenes, titulos):
        ax.imshow(img)
        ax.set_title(titulo)
        ax.axis('off')
        plt.tight_layout()
    plt.show()
```

```python
#metodo para mostrar canales greyscale
def mostrar_canales(img, titulos=("Canal 1", "Canal 2", "Canal 3")):
    A,B,C = cv.split(img)
    canales = [A,B,C]

    fig, axes = plt.subplots(1,3,figsize=(5,6))
    for(ax, img, titulo) in zip(axes, canales, titulos):
        ax.imshow(img, cmap = 'gray')
        ax.set_title(titulo)
        ax.axis('off')
        plt.tight_layout()
    plt.show()
   
```

```python
bright = cv.imread('../data/raw/agua/agua_brillo.jpg')
dark = cv.imread('../data/raw/agua/agua_oscuro.jpg')
bright_RGB = cv.cvtColor(bright, cv.COLOR_BGR2RGB)
dark_RGB = cv.cvtColor(dark,cv.COLOR_BGR2RGB)
mostrar_dos_imagenes(bright_RGB, dark_RGB)
```

```python
mostrar_canales(bright_RGB)
mostrar_canales(dark_RGB)
```

```python
brightLAB = cv.cvtColor(bright, cv.COLOR_BGR2LAB)
darkLAB = cv.cvtColor(dark, cv.COLOR_BGR2LAB)
mostrar_dos_imagenes(brightLAB, darkLAB)
```

```python
mostrar_canales(brightLAB)
mostrar_canales(darkLAB)
```

```python
brightYCB = cv.cvtColor(bright, cv.COLOR_BGR2YCrCb)
darkYCB = cv.cvtColor(dark, cv.COLOR_BGR2YCrCb)
mostrar_dos_imagenes(brightYCB,darkYCB)
```

```python
mostrar_canales(brightYCB)
mostrar_canales(darkYCB)
```

```python
brightHSV = cv.cvtColor(bright, cv.COLOR_BGR2HSV)
darkHSV = cv.cvtColor(dark, cv.COLOR_BGR2HSV)
mostrar_dos_imagenes(brightHSV,darkHSV)
```

```python
mostrar_canales(brightHSV)
mostrar_canales(darkHSV)
```

```python


# Listas temporales para acumular canales (mucho más rápido que np.append)
b_list, g_list, r_list = [], [], []
h_list, s_list, v_list = [], [], []
l_list, a_list, b_lab_list = [], [], []
y_list, cr_list, cb_list = [], [], []

for fi in image_paths:
    # 1. Leer imagen en BGR original
    mimami = cv.imread(fi)
    if mimami is None:
        continue

    # Canales BGR directos (0=B, 1=G, 2=R)
    b_list.append(mimami[:, :, 0].flatten())
    g_list.append(mimami[:, :, 1].flatten())
    r_list.append(mimami[:, :, 2].flatten())

    # 2. HSV (0=Hue, 1=Saturation, 2=Value)
    im_hsv = cv.cvtColor(mimami, cv.COLOR_BGR2HSV)
    h_list.append(im_hsv[:, :, 0].flatten())
    s_list.append(im_hsv[:, :, 1].flatten())
    v_list.append(im_hsv[:, :, 2].flatten())

    # 3. CIELAB (0=L*, 1=a*, 2=b*)
    im_lab = cv.cvtColor(mimami, cv.COLOR_BGR2LAB)
    l_list.append(im_lab[:, :, 0].flatten())
    a_list.append(im_lab[:, :, 1].flatten())
    b_lab_list.append(im_lab[:, :, 2].flatten())

    # 4. YCrCb (0=Y, 1=Cr, 2=Cb)
    im_ycrcb = cv.cvtColor(mimami, cv.COLOR_BGR2YCrCb)
    y_list.append(im_ycrcb[:, :, 0].flatten())
    cr_list.append(im_ycrcb[:, :, 1].flatten())
    cb_list.append(im_ycrcb[:, :, 2].flatten())

# Concatenar todos los píxeles acumulados en arreglos 1D finales
B = np.concatenate(b_list)
G = np.concatenate(g_list)
R = np.concatenate(r_list)

H = np.concatenate(h_list)
S = np.concatenate(s_list)
V = np.concatenate(v_list)

L = np.concatenate(l_list)
A = np.concatenate(a_list)
B_lab = np.concatenate(b_lab_list)

Y = np.concatenate(y_list)
Cr = np.concatenate(cr_list)
Cb = np.concatenate(cb_list)
```
